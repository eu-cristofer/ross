Vary `nfft`, overlap, and segment count (via record length) to inspect stability of `B11`, `B22`, and `T111` magnitudes on a fixed synthetic or rotor-derived segment.

## References

Sinha, J. K. (2007). Higher order spectra for crack and misalignment identification in the shaft of a rotating machine. *Structural Health Monitoring*, 6(4), 325-334.


```python
import sys
import numpy as np
import pandas as pd
from ross.hos.bispectrum import estimate_bispectrum

if "." not in sys.path:
    sys.path.insert(0, ".")
import sinha_tools as st

fs = 256.0
t = np.arange(0, 30, 1 / fs)
fr = 12.5
x = np.cos(2 * np.pi * fr * t) + np.cos(2 * np.pi * 2 * fr * t) + 0.4 * np.cos(2 * np.pi * 3 * fr * t)
rows = []
for nfft in (512, 1024, 2048):
    noverlap = nfft // 2
    d = st.fault_indicators(x, fs, fr, nfft=nfft, noverlap=noverlap)
    d["nfft"] = nfft
    rows.append(d)
print(pd.DataFrame(rows))
```
