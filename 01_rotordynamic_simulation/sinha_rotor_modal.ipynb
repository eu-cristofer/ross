```python
import numpy as np
import ross as rs
from ross.materials import Material

# Make sure the default renderer is set to 'notebook' for inline plots in Jupyter
import plotly.io as pio

pio.renderers.default = "notebook"

print(f"ROSS version: {rs.__version__}")

Q_ = rs.Q_

rotor = rs.Rotor.load("sinha_rotor.toml")
```

```python
modal = rotor.run_modal(speed=Q_(100, "rad/s"), num_modes=12)

print("\nUndamped natural frequencies (Hz):")
print((modal.wn / (2 * np.pi)))

print("\nDamped natural frequencies (Hz):")
print((modal.wd / (2 * np.pi)))

print("\nDamping ratios:")
print(modal.damping_ratio)
```

```python
modal.plot_mode_3d(mode=1, frequency_units="Hz")
```

## Unbalance response

```python
# Define unbalance
node = [6]
unbalance_magnitude = [1.0e-4]
unbalance_phase = [0]

# Speed range
speed = Q_(np.linspace(0, 6000, 200), "RPM")

# Run analysis
resp = rotor.run_unbalance_response(
    node=node,
    unbalance_magnitude=unbalance_magnitude,
    unbalance_phase=unbalance_phase,
    speed_range=speed
)

# Plot results
resp.plot()
```
