# 11 - Fault discrimination (scalar rules)

Feature table from `fault_indicators` on archived HDF5 traces (Section 06) or short re-simulations. Rules are deterministic: elevated `B22` supports crack hypothesis; elevated `B13` supports misalignment hypothesis; `T222` relative to `T111` is reported. No machine-learning classifier is used.

## References

Sinha, J. K. (2007). Higher order spectra for crack and misalignment identification in the shaft of a rotating machine. *Structural Health Monitoring*, 6(4), 325-334.


```python
import sys
from pathlib import Path
import pandas as pd

if "." not in sys.path:
    sys.path.insert(0, ".")
import sinha_tools as st

rows = []
hdir = Path("study_docs/sinha_validation/hdf5")
if hdir.exists():
    for p in sorted(hdir.glob("*.h5")):
        if "_test" in p.name:
            continue
        t, y, fs, meta = st.load_timeseries_hdf5(p)
        rpm = float(meta.get("rpm", 0))
        fr = rpm / 60.0
        d = st.fault_indicators(y, fs, fr)
        d.update(meta)
        rows.append(d)
if rows:
    print(pd.DataFrame(rows))
else:
    print("No HDF5 archives found; run 06 first.")
```
