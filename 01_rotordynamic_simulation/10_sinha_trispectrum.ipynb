# 10 - Trispectrum slices (Sinha Figs. 7-8)

Only the slice magnitudes requested in the validation plan are computed (`ross.hos.estimate_trispectrum_point`).

## References

Sinha, J. K. (2007). Higher order spectra for crack and misalignment identification in the shaft of a rotating machine. *Structural Health Monitoring*, 6(4), 325-334.


```python
import numpy as np
import sys
import pandas as pd

if "." not in sys.path:
    sys.path.insert(0, ".")
import sinha_tools as st

rotor = st.load_sinha_rotor("sinha_rotor.toml")
probe = st.default_probe(rotor)
fs0 = 2560.0
dt = 1.0 / fs0
t = np.arange(0, int(10 * fs0)) * dt
rpm = 750
speed = rpm * 2 * np.pi / 60.0
fr = rpm / 60.0

rows = []
for case in ("crack", "misalignment"):
    res = st.simulate_case(rotor, case, speed, t, num_modes=20)
    y = st.probe_displacement_y_from_rotor(rotor, res, probe)
    yp, fs = st.acquisition_chain(y[int(2 * fs0) :], fs0)
    d = st.fault_indicators(yp, fs, fr, nfft=2048, noverlap=1024)
    d["case"] = case
    rows.append(d)
print(pd.DataFrame(rows))
```
